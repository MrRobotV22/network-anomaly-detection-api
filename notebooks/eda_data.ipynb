```python
def detect_outliers(col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    print(f"{col}: {len(outliers)} outliers")

for col in ['srcbytes', 'dstbytes', 'count' , 'serrorrate' , 'duration']:
    detect_outliers(col)
```

## Hypothesis Testing

### Hypothesis 1: Traffic Asymmetry (src_bytes vs dst_bytes)

Due to skewed distribution and presence of extreme values, Let's use a non-parametric Mann–Whitney U test instead of a t-test(t-test assumes normal distribution and looks at Mean only not at distribution).
- H₀ (Null): There is no difference in byte ratio between normal and anomalous connections
- H₁ (Alt): Anomalous connections have significantly different byte ratios

```python
from scipy.stats import mannwhitneyu

df['byte_ratio'] = df['srcbytes'] / (df['dstbytes'] + 1)

normal = df[df['Anomaly'] == 0]['byte_ratio']
anomaly = df[df['Anomaly'] == 1]['byte_ratio']

stat, p_value = mannwhitneyu(normal, anomaly, alternative='two-sided')

print("Mann-Whitney U Statistic:", stat)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Result: Reject H0 → Significant difference in byte_ratio")
else:
    print("Result: Fail to reject H0 → No significant difference")
```

```python
print("Median byte_ratio (Normal):", normal.median())
print("Median byte_ratio (Anomaly):", anomaly.median())

# Effect size (rank-biserial correlation)
n1 = len(normal)
n2 = len(anomaly)

effect_size = 1 - (2 * stat) / (n1 * n2)
print("Effect Size (Rank-Biserial):", effect_size)
```

- Anomalous connections exhibit significantly lower byte ratios compared to normal traffic (median: 0.0 vs 0.346), indicating highly asymmetric communication patterns.
- The large effect size (-0.59) confirms that this difference is not only statistically significant but also practically meaningful, suggesting behaviors such as failed connections, probing, or attack attempts with minimal data exchange.

### Hypothesis 2: Failed Login Behavior

- H₀ (Null): Failed login behavior is independent of anomalies
- H₁ (Alt): Failed login behavior is associated with anomalies

```python
from scipy.stats import chi2_contingency

df['failedloginflag'] = ((df['numfailedlogins'] > 0) & (df['loggedin'] == 0)).astype(int)

contingency_table = pd.crosstab(df['failedloginflag'], df['Anomaly'])
print("Contingency Table:\n", contingency_table)
```

```python
prob = contingency_table.div(contingency_table.sum(axis=1), axis=0)
print(prob)
```

```python
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("\nChi-square Statistic:", chi2)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Result: Reject H0 → Failed login behavior is associated with anomalies")
else:
    print("Result: Fail to reject H0 → No significant association")
```

Although the Chi-square test did not show statistical significance (p = 0.056), conditional probability analysis reveals that connections with failed login attempts have a higher likelihood of being anomalous (56.8% vs 46.5%). This suggests that failed login behavior, while rare, is a meaningful indicator of suspicious activity.

### Hypothesis 3: Connection Burst (count)

- H₀ (Null): No difference in count between normal and anomalous connections
- H₁ (Alt): Anomalous connections have significantly different (higher) count

```python
normal = df[df['Anomaly'] == 0]['count']
anomaly = df[df['Anomaly'] == 1]['count']

print("Median count (Normal):", normal.median())
print("Median count (Anomaly):", anomaly.median())
n1 = len(normal)
n2 = len(anomaly)
effect_size = 1 - (2 * stat) / (n1 * n2)
print("Effect Size:", effect_size)

stat, p_value = mannwhitneyu(normal, anomaly, alternative='two-sided')

print("Mann-Whitney U Statistic:", stat)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Result: Reject H0 → Significant difference in connection count")
else:
    print("Result: Fail to reject H0 → No significant difference")
```

Anomalous connections exhibit significantly higher connection counts within short time windows (median: 138 vs 4), indicating burst behavior. The large effect size (0.64) confirms a strong separation between normal and anomalous traffic. This pattern is characteristic of automated attacks such as DDoS and port scanning, making 'count' one of the most important features for anomaly detection.
